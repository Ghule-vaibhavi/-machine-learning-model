naive bayes , fruit classifier

```python
from sklearn.naive_bayes import CategoricalNB

# Color: Red=0, Green=1, Orange=2
# Shape: Round=0

X = [
    [0, 0],  # Red, Round
    [0, 0],  # Red, Round
    [1, 0],  # Green, Round
    [2, 0],  # Orange, Round
    [2, 0],  # Orange, Round
    [1, 0]   # Green, Round
]

# Apple=0, Orange=1
y = [0, 0, 0, 1, 1, 1]

model = CategoricalNB(alpha=1)
model.fit(X, y)

# Predict a new red, round fruit
prediction = model.predict([[0, 0]])

if prediction[0] == 0:
    print("Apple")
else:
    print("Orange")
```
